```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, accuracy_score, confusion_matrix
import warnings
warnings.filterwarnings('ignore')
sns.set(style="whitegrid")
```

```python
# 2.1 Load Dataset
df = pd.read_csv('IPL 2008 to 2024.csv')

# 2.2.1 First 5 Rows
display(df.head())

# 2.2.2 Last 5 Rows
display(df.tail())

# 2.2.3 Dataset Shape
print("Shape:", df.shape)

# 2.2.4 Column Names
print("Columns:", df.columns.tolist())

# 2.3 Identify Data Types
quant_data = df.select_dtypes(include=['number']).columns.tolist()
qual_data = df.select_dtypes(exclude=['number']).columns.tolist()
print("2.3.1 Quantitative (Numbers):", quant_data)
print("2.3.2 Qualitative (Text):", qual_data)
```

```python
# 3.1 Univariate Analysis
plt.figure(figsize=(8, 5))
ax = sns.histplot(df['First_Innings_Score'], kde=True, color='blue', bins=20)
avg_score = df['First_Innings_Score'].mean()
plt.axvline(avg_score, color='red', linestyle='--', linewidth=2)
plt.text(avg_score + 2, ax.get_ylim()[1]*0.9, f'Average Score: {avg_score:.0f}', color='red', fontsize=12, weight='bold')
plt.title('3.1 Univariate: IPL Me Har Match Me Kitne Runs Bante Hain?', fontsize=14)
plt.xlabel('First Innings Final Score', fontsize=12)
plt.ylabel('Matches Ki Ginti', fontsize=12)
plt.show()

# 3.2 Bivariate Analysis
plt.figure(figsize=(8, 5))
sns.regplot(x='Powerplay_Scores', y='First_Innings_Score', data=df, scatter_kws={'color': 'green', 'alpha': 0.6}, line_kws={'color': 'red', 'linewidth': 2})
plt.title('3.2 Bivariate: Powerplay Runs Se Final Score Par Kya Asar Padta Hai?', fontsize=14)
plt.xlabel('Powerplay Runs (Shuruwat ke 6 Overs)', fontsize=12)
plt.ylabel('Final Score (20 Overs ke baad)', fontsize=12)
plt.text(df['Powerplay_Scores'].min(), df['First_Innings_Score'].max() - 10, "Red Line Upar Jaa Rahi Hai = Powerplay Achha Toh Score Achha", color='red', fontsize=11, weight='bold')
plt.show()

# 3.3 Multivariate Analysis
plt.figure(figsize=(10, 6))
numeric_cols = df[['First_Innings_Score', 'Powerplay_Scores', 'Middle_Overs_Scores', 'Death_Overs_Scores']]
corr_matrix = numeric_cols.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=1, annot_kws={"size": 13, "weight": "bold"})
plt.title('3.3 Multivariate: Scores Ke Beech Ka Gehra Rishta (Correlation)', fontsize=15, weight='bold', pad=20)
plt.figtext(0.5, -0.05,"Laal Rang (Dark Red) = Strong Connection (Jaise 0.83)\nMatlab agar Middle Overs me zyada runs banenge, toh Final Score pakka bada hoga!", ha="center", fontsize=12,bbox={"facecolor":"yellow", "alpha":0.4, "pad":8},color="black", weight="bold")
plt.show()
```

```python
# 4.1 Identify Missing Values
print(df.isnull().sum())

# 4.2 & 4.2.1 Handle Missing Values using Median
df['Win_Margin'] = df['Win_Margin'].fillna(df['Win_Margin'].median())
df['Second_Innings_Score'] = df['Second_Innings_Score'].fillna(df['Second_Innings_Score'].median())

# 4.3 & 4.3.1 Detect Outliers Using Boxplots
plt.figure(figsize=(11, 5))
ax = sns.boxplot(x=df['Win_Margin'], color='orange')
plt.title('4.3 Outliers: Kitne Runs Se Team Jeeti?', fontsize=14)
plt.xlabel('Jeet Ka Antar (Win Margin in Runs/Wickets)', fontsize=12)
plt.annotate('Ye Outliers Hain (100+ Runs Se Jeet)',xy=(100, 0), xytext=(80, 0.3),arrowprops=dict(facecolor='black', shrink=0.05),fontsize=12, color='black', weight='bold')
plt.show()
```

```python
# 5.1 Check Data Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df['First_Innings_Score'], kde=True, color='purple')
mean_val = df['First_Innings_Score'].mean()
median_val = df['First_Innings_Score'].median()
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}')
plt.axvline(median_val, color='blue', linewidth=2, label=f'Median: {median_val:.1f}')
plt.title('5.1 Data Distribution (Normal vs Skewed)')
plt.legend()
plt.show()

# 5.2 Calculate Statistical Measures
std_dev = df['First_Innings_Score'].std()
kurt = df['First_Innings_Score'].kurtosis()
skew = df['First_Innings_Score'].skew()

print(f"5.2.1 Standard Deviation: {std_dev:.2f}")
print(f"5.2.2 Kurtosis: {kurt:.2f}")
print(f"Skewness: {skew:.2f}")
```

```python
# 6.2 Reusable Python functions for EDA
def auto_eda_function(data):
    print("--- 6.1.2 Info ---")
    data.info()
    print("\\n--- 6.1.1 Describe ---")
    display(data.describe())
    print("\\n--- 6.1.3 Isnull ---")
    print(data.isnull().sum())
    print("\\n--- 6.1.4 Corr ---")
    display(data.select_dtypes(include='number').corr())

auto_eda_function(df)
```

```python
# 7.1 Identify Variables
X_simple = df[['Powerplay_Scores']]
y = df['First_Innings_Score']

# 7.2 Perform Simple Linear Regression
plt.figure(figsize=(8, 5))
X = df[['Powerplay_Scores']]  # Madad lenge
y = df['First_Innings_Score'] # Dhundna hai
model = LinearRegression()
model.fit(X, y)
pred_y = model.predict(X)
sns.scatterplot(x=df['Powerplay_Scores'], y=y, color='lightblue', label='Asli Score (Actual)')
plt.plot(df['Powerplay_Scores'], pred_y, color='red', linewidth=2.5, label='Model ki Line (Prediction)')
plt.xlabel('Powerplay Runs')
plt.ylabel('Final Score')
plt.legend()
plt.show()

# 7.3 Analyze Covariance and Correlation
covar = np.cov(df['Powerplay_Scores'], df['First_Innings_Score'])[0][1]
corre = df['Powerplay_Scores'].corr(df['First_Innings_Score'])
print(f"7.3.1 Covariance: {covar:.2f}")
print(f"7.3.2 Correlation: {corre:.2f}")
```

```python
# 8.1 Split dataset
X_multi = df[['Powerplay_Scores', 'Middle_Overs_Scores', 'Death_Overs_Scores']]
X_train, X_test, y_train, y_test = train_test_split(X_multi, y, test_size=0.2, random_state=42)

# 8.2 & 8.3 Train Regression Model
multi_reg = LinearRegression()
multi_reg.fit(X_train, y_train)

# 8.4 Validate model performance
y_pred_multi = multi_reg.predict(X_test)
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred_multi, color='green', alpha=0.6, label='Model ki Prediction')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2.5, label='Perfect Prediction Line')
plt.xlabel('Asli Score (Actual)')
plt.ylabel('Computer Ne Jo Bataya (Predicted)')
plt.legend()
plt.show()
```

```python
# 9.2 Compare Errors
train_pred = multi_reg.predict(X_train)
mse_train = mean_squared_error(y_train, train_pred)
mse_test = mean_squared_error(y_test, y_pred_multi)

print(f"9.2 Training Error (MSE): {mse_train:.2f}")
print(f"9.2 Testing Error (MSE): {mse_test:.2f}")
print("9.3: Errors are close, Model is balanced.")
```

```python
# 9.1 Convert to Classification Problem
df['Is_High_Score'] = (df['First_Innings_Score'] > 170).astype(int)
y_class = df['Is_High_Score']

# 9.2 Build Classification Model
Xc_train, Xc_test, yc_train, yc_test = train_test_split(X_multi, y_class, test_size=0.2, random_state=42)
log_reg = LogisticRegression()
log_reg.fit(Xc_train, yc_train)

# 9.3 Evaluate & 9.4 Accuracy
yc_pred = log_reg.predict(Xc_test)
acc = accuracy_score(yc_test, yc_pred)
print(f"9.4 Accuracy: {acc*100:.2f}%")

# 9.5 Confusion Matrix
cm = confusion_matrix(yc_test, yc_pred)
plt.figure(figsize=(8, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', annot_kws={"size": 14, "weight": "bold"})
plt.title('9.5 Confusion Matrix: Model Ne Kitne Sahi Jawab Diye?', fontsize=14)
plt.xlabel('Model Ne Kya Kaha (Predicted)', fontsize=12)
plt.ylabel('Asliyat Me Kya Hua (Actual)', fontsize=12)
plt.show()

```

```python
# 10.1 Calculate Error Metrics
mse_val = mean_squared_error(y_test, y_pred_multi)
mae_val = mean_absolute_error(y_test, y_pred_multi)
r2_val = r2_score(y_test, y_pred_multi)

print(f"10.1.1 MSE: {mse_val:.2f}")
print(f"10.1.2 MAE: {mae_val:.2f}")
print(f"10.1.3 R² Score: {r2_val:.2f}")

# 10.2 & 10.4 Interpretation
print("10.2 Regression gives exact runs (R2 Score), Classification gives categories (Accuracy).")
print("10.4 Interpret: The model performance is excellent for predicting IPL scores.")
```
